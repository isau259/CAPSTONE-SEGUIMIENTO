```python
def deteccion(sub, ses, run, seg):
    
    # 1. Crear rutas a los archivos
    datos = fn2.crear_variable(sub, ses, run, seg)
    
    # 2. Procesar la señal (con la config de step=5)
    df = fn2.procesamiento(datos, win_size=150, step=5, smooth_n=5, overview_ds=10)
    
    # 3. Calcular alertas
    df_alertas = fn2.calcular_baseline_y_alertas(
        df,
        window_rolling=300,  
        persistence_sec=10,
        min_votes=3
    )
    
    # 4. Visualizar (¡AÑADE LAS NUEVAS MÉTRICAS AQUÍ!)
    fn2.viewer_plotly_params_from_df(
        df_alertas,
        metrics=(
            "ALERTA_CONSENSO",
            "ALERTA_PERSISTENTE",
            "HR_mean",
            "SDNN_ms",
            "LF/HF",
            # --- NUEVAS MÉTRICAS A VISUALIZAR ---
            "CSI",
            "ModCSI",
            "Tachogram_Slope",
            "CSI_Slope",
            "ModCSI_Slope"
            # -----------------------------------
        ),
        default_metric="ALERTA_PERSISTENTE",
        default_smooth=1,
        open_in_browser=False
    )


deteccion(sub=2, ses=1, run=9, seg=0)
```
